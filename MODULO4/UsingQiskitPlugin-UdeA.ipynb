```python
import pennylane as qml
from pennylane import numpy as np
```

```python
from qiskit_aer import Aer
```

```python
# Create a quantum device with 1 qubit
dev1 = qml.device("default.qubit", wires=1)
dev2 = qml.device("qiskit.aer", wires=1)
```

```python
def Ansatz(theta):
    qml.RY(theta, wires=0)
```

```python
# Define the variational quantum circuit
@qml.qnode(dev1)
def circuit1(theta):
    Ansatz(theta)

    # Measurement
    return qml.expval(qml.PauliZ(0))
```

```python
# Define the variational quantum circuit
@qml.qnode(dev2)
def circuit2(theta):
    Ansatz(theta)

    # Measurement
    return qml.expval(qml.PauliZ(0))
```

```python
print("Expected value using the default.qubit device:")
print(circuit1(0.5))

print("Expected value using the qiskit.aer device:")
print(circuit2(0.5))
```

```python
qml.drawer.use_style("pennylane")
qml.draw_mpl(circuit1, level=2)(0.5)
```

```python
dev2._circuit.draw(output='mpl')
```

```python
# Cost function
def cost(theta):
    return circuit2(theta)
```

```python
# Initial parameter
theta = np.array(0.1, requires_grad=True)
```

```python
# Classical optimizer
optimizer = qml.GradientDescentOptimizer(stepsize=0.2)
```

```python
# Optimization loop
steps = 50

for i in range(steps):

    theta = optimizer.step(cost, theta)

    if (i + 1) % 5 == 0:
        print(f"Step {i+1}:")
        print("Theta =", theta)
        print("Cost =", cost(theta))
        print()
```

```python
print("Final optimized theta:", theta)
print("Final minimum cost:", cost(theta))
```
